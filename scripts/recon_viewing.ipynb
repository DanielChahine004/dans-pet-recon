```python
import panel as pn
import matplotlib.pyplot as plt
pn.extension()

positions = [0, 32, 64, 96, 120]

xy_slider = pn.widgets.DiscreteSlider(name='XY', options=positions, value=0)
z_slider  = pn.widgets.DiscreteSlider(name='Z',  options=positions, value=0)
filter_slider = pn.widgets.DiscreteSlider(name='LOR Filter', options=['none', 'midway'], value='none')
@pn.depends(xy_slider, z_slider, filter_slider)
def update(xy, z, filter_type):
    source_pos = (xy, xy, z)

    if filter_type == 'none':
        path = '/home/h/Code/Opengate/ragy_files/Point_Sources_Processed'
    else:
        path = '/home/h/Code/Opengate/ragy_files/Point_Sources_r235_to_257mm'
    with open(f'{path}/{source_pos[0]}_{source_pos[1]}_{source_pos[2]}/gt/recon_config.pkl', "rb") as f:
        config = pickle.load(f)

    gt_image = np.load(f'{path}/{source_pos[0]}_{source_pos[1]}_{source_pos[2]}/gt/reconstruction_image.npy')
    prediction_image = np.load(f'{path}/{source_pos[0]}_{source_pos[1]}_{source_pos[2]}/pred/reconstruction_image.npy')

    if SMOOTH_RECON_FOR_DISPLAY:
        gt_image = gaussian_filter(gt_image, sigma=DISPLAY_SMOOTH_SIGMA, radius=DISPLAY_SMOOTH_RADIUS)
        prediction_image = gaussian_filter(prediction_image, sigma=DISPLAY_SMOOTH_SIGMA, radius=DISPLAY_SMOOTH_RADIUS)

    hottest_gt_x, hottest_gt_y, hottest_gt_z = np.unravel_index(np.argmax(gt_image), gt_image.shape)
    hottest_pred_x, hottest_pred_y, hottest_pred_z = np.unravel_index(np.argmax(prediction_image), prediction_image.shape)

    hottest_gt_x_coord   = hottest_gt_x   - config["IMAGE_GRID_SIZE"][0]//2 * config["VOXEL_SIZE_IN_MM"]
    hottest_gt_y_coord   = hottest_gt_y   - config["IMAGE_GRID_SIZE"][1]//2 * config["VOXEL_SIZE_IN_MM"]
    hottest_gt_z_coord   = hottest_gt_z   - config["IMAGE_GRID_SIZE"][2]//2 * config["VOXEL_SIZE_IN_MM"]
    hottest_pred_x_coord = hottest_pred_x - config["IMAGE_GRID_SIZE"][0]//2 * config["VOXEL_SIZE_IN_MM"]
    hottest_pred_y_coord = hottest_pred_y - config["IMAGE_GRID_SIZE"][1]//2 * config["VOXEL_SIZE_IN_MM"]
    hottest_pred_z_coord = hottest_pred_z - config["IMAGE_GRID_SIZE"][2]//2 * config["VOXEL_SIZE_IN_MM"]

    gt_dist   = np.sqrt((hottest_gt_x_coord  - source_pos[0])**2 + (hottest_gt_y_coord  - source_pos[1])**2 + (hottest_gt_z_coord  - source_pos[2])**2)
    pred_dist = np.sqrt((hottest_pred_x_coord - source_pos[0])**2 + (hottest_pred_y_coord - source_pos[1])**2 + (hottest_pred_z_coord - source_pos[2])**2)

    info = pn.pane.Str(
        f"Source position: {source_pos}\n"
        f"Hottest GT voxel:   ({hottest_gt_x_coord}, {hottest_gt_y_coord}, {hottest_gt_z_coord})  →  {gt_dist:.2f}mm  {'✅' if gt_dist < 3.0 else '❌'}\n"
        f"Hottest PRED voxel: ({hottest_pred_x_coord}, {hottest_pred_y_coord}, {hottest_pred_z_coord})  →  {pred_dist:.2f}mm  {'✅' if pred_dist < 3.0 else '❌'}"
    )

    plot_static(gt_image, title="gt_image", axis=2, depth=hottest_gt_z, voxel_size=config['VOXEL_SIZE_IN_MM'])
    fig1 = plt.gcf(); plt.close()

    plot_static(prediction_image, title="prediction_image", axis=2, depth=hottest_gt_z, voxel_size=config['VOXEL_SIZE_IN_MM'])
    fig2 = plt.gcf(); plt.close()

    plot_line_profile_with_gaussian_fit(gt_image, axis='z', coord1=hottest_gt_x, coord2=hottest_gt_y, voxel_size=config["VOXEL_SIZE_IN_MM"], show=False)
    fig3 = plt.gcf(); plt.close()

    plot_line_profile_with_gaussian_fit(prediction_image, axis='z', coord1=hottest_gt_x, coord2=hottest_gt_y, voxel_size=config["VOXEL_SIZE_IN_MM"], show=False)
    fig4 = plt.gcf(); plt.close()

    return pn.Column(
        info,
        pn.Row(
            pn.pane.Matplotlib(fig1, tight=True, sizing_mode='stretch_width'),
            pn.pane.Matplotlib(fig2, tight=True, sizing_mode='stretch_width'),
            sizing_mode='stretch_width',
        ),
        pn.Row(
            pn.pane.Matplotlib(fig3, tight=True, sizing_mode='stretch_width'),
            pn.pane.Matplotlib(fig4, tight=True, sizing_mode='stretch_width'),
            sizing_mode='stretch_width',
        ),
        sizing_mode='stretch_width',
    )

pn.Column(xy_slider, z_slider, filter_slider, update)
```
